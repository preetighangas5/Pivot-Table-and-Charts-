**Question 4:** Explain the difference between Sorting and Filtering with an example from this dataset.

**Answer: ** SORTING-

*   Sorting means rearranging the data in a specific order (ascending or descending based on column values.
*   In this all data remains visible, just in a new order.

     FILTERING-

*   Filtering means temporary hides data rows that do not match the selected criteria, leaving only relevant rows visible.
*  Rows that do not meet the criteria are hidden.

 Here in this example Sorting means arrange sales values from largest to smallest to identify top performing sales. Filtering means show only data related to like mangoes to analyze mango sales.

**Question 7:** Why are Dashboards important for decision-making in business analysis?

**Answer:** Dashboards provide a visual summary of key performance indicators (KPIs), helping decision-makers analyze trends, patterns, and performance quickly.
 Dashboards are inportant for decision making in business analysis because

*   Save time by showing insights at a glance
*   Improve accuracy in decision making

*   Help track performance and trends
*   Support strategic planning
